```python
import pandas as pd
```

```python
file_path = "C:/Users/jiele/OneDrive - Universiti Malaya/WQD 7006 MACHINE LEARNING/Raw dataset/gdp_district_real_supply.csv"
df = pd.read_csv(file_path)
df
```

```python
# Remove rows where the 'series' column has the value 'growth_yoy'
df = df[df['series'] != 'growth_yoy']
check = df['series'].unique()
print(check)
```

```python
#drop column series
df =df.drop(columns=['series'])
print(df)
```

```python
# Convert 'date' Column to Year Only
df['date'] = pd.to_datetime(df['date']).dt.year
```

```python
df
```

```python
##transpose the dataset
df2 = df.pivot(index=['date','state','district'], columns='sector', values='value').reset_index()
```

```python
 # Remove the header name
df2.columns.name = None 
```

```python
#rearange the columns
df2 =df2[['date','state','district','p0','p1','p2','p3','p4','p5']]
```
