# Prepare for Merging

```python
df_crime = pd.read_csv('../data/massaged/crime.csv')
df_gdp = pd.read_csv('../data/massaged/gdp.csv')
```

```python
df_crime.columns
```

```python
df_gdp.columns
```

## Merge Johor Bahru Utara and Selatan into one

```python
print(df_crime[df_crime['district'].isin(['Johor Bahru Utara', 'Johor Bahru Selatan'])])
```

```python
df_crime['district'] = df_crime['district'].replace(['Johor Bahru Utara', 'Johor Bahru Selatan'], 'Johor Bahru')
df_crime = df_crime.groupby(['year', 'district']).agg({'crimes': 'sum'}).reset_index()
```

```python
df_crime_johor_bahru = df_crime[df_crime['district'] == 'Johor Bahru']
print(df_crime_johor_bahru)
```

## Merge Klang Utara and Selatan into one

```python
print(df_crime[df_crime['district'].isin(['Klang Utara', 'Klang Selatan'])])
```

```python
df_crime['district'] = df_crime['district'].replace(['Klang Utara', 'Klang Selatan'], 'Klang')
df_crime = df_crime.groupby(['year', 'district']).agg({'crimes': 'sum'}).reset_index()
```

```python
df_crime_klang = df_crime[df_crime['district'] == 'Klang']
print(df_crime_klang)
```

## Convert Petaling Jaya to Petaling

```python
df_crime['district'] = df_crime['district'].replace('Petaling Jaya', 'Petaling')
```

## Remove Districts not Found in GDP data

```python
removed_districts = df_crime[~df_crime['district'].isin(df_gdp['district'].unique())]['district'].unique()
df_crime = df_crime[df_crime['district'].isin(df_gdp['district'].unique())]
print(f"The following districts were removed: {removed_districts}")
```

```python
df_gdp.district.unique()
```

```python
df_gdp = df_gdp.drop('state', axis=1)
```

# Merge

```python
(df_gdp.isna().sum() / len(df_gdp)) * 100
```

```python
df_merged = df_crime.merge(df_gdp, on=['district', 'year'], how='inner')
df_merged.isna().sum()
```

```python
df_merged.to_csv('gdp_and_crime.csv', index=False)
```
